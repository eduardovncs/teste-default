Cobre R5-18 (operadores multilinha `!=` e `%`) e R5-22 (`%timeit` de linha única).

```python
status = 'ATIVO'
total = 125

# R5-18: estilo PEP 8 / Black / Ruff com operador no inicio da linha
ativo = (
    status
    != 'CANCELADO'
)
resto = (
    total
    % 10
)

df = spark.table('dev.src.tb_vendas')
if ativo and resto == 5:
    df.write.mode('overwrite').saveAsTable('dev.tgt.tb_nb_r5_ops')
```

```python
# R5-22: %timeit de linha unica
%timeit df.write.mode('overwrite').saveAsTable('dev.tgt.tb_nb_r5_timeit')
```
